# Orderrapport i jobbläge

Den här notebooken visar projektets flöde steg för steg. Funktionerna ligger i paketet `order_report`, så samma kod används här, i `main.py` och i testerna. Kör cellerna uppifrån och ned.

```python
import logging

from order_report import (
    ReportConfig, add_order_values, build_reports, clean_orders,
    load_orders, summarize_orders
)
from order_report.reporting import save_reports

logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
config = ReportConfig()
config
```

## 1 Läs in och validera data
`load_orders` kontrollerar filen, obligatoriska kolumner och att datasetet innehåller rader.

```python
raw_orders = load_orders(config)
raw_orders.head()
```

## 2 Tvätta data
Textvärden normaliseras, numeriska kolumner görs om till tal och returvärden blir `True` eller `False`.

```python
cleaned_orders = clean_orders(raw_orders)
cleaned_orders.head()
```

## 3 Beräkna ordervärden
`order_value` är antal gånger styckpris. `discounted_value` tar även hänsyn till rabatten.

```python
orders = add_order_values(cleaned_orders)
orders[['order_id', 'quantity', 'unit_price', 'discount', 'order_value', 'discounted_value']].head()
```

## 4 Skapa sammanfattning och rapporter

```python
summary = summarize_orders(orders)
summary
```

```python
reports = build_reports(orders, summary)
reports['sales_by_category.csv']
```

## 5 Spara rapporterna
Fyra CSV-filer sparas i mappen `output`.

```python
saved_paths = save_reports(reports, config)
saved_paths
```

Testerna kontrollerar beräkningar, datatvätt, rapporter och felfall.
